```python
class RemoteControl:
    def __init__(self):
        self.channels = ["NDTV","Disney","HBO"]
        self.index = -1

    def __next__(self):
        self.index += 1
        if self.index == len(self.channels):
            raise StopIteration
        else:
            return self.channels[self.index]
    def __iter__(self):
        return self
```

```python
rc = RemoteControl()
next(rc)
```

```python
def fibbo():
    a,b = 0,1
    while True:
        yield a
        a,b = b, a+b
```

```python
fib = fibbo()
```

```python
for _ in range(10):
    print(next(fib))
```
