```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("miles.csv")
df.head()
```

Let's calculate the population mean and std dev from which we will have std dev only mean won't be available to us in real life. But here we are trying to prove the C.I Formula.

```python
df.miles.mean()
```

```python
std = df.miles.std()
std
```

```python
plt.figure(figsize=(12,4))
plt.plot(df.miles)
```

```python
sns.histplot(df.miles, kde = True)
```

As we can see it is right skewed distribution

Now to prove the formula of confidence interval let's take a random sample of size 50

```python
SAMPLE_SIZE = 50
```

Now using .sample method we will take any random sample values from our population. Here for reproducibality we will pass some random_state to it

```python
sample_data = df.sample(SAMPLE_SIZE, random_state = 100)
sample_data.head()
```

Now let's compute the sample mean

```python
sample_mean = sample_data.miles.mean()
sample_mean
```

Now let's find std error using the population std dev and sample size.

```python
SE = std/np.sqrt(SAMPLE_SIZE)
SE
```

Now let's find Z value at 95% Confidence Level using scipy.states ppf method
= 95% + 2.5% = 97.5%

```python
import scipy.stats as st
z = st.norm.ppf(0.975)
z
```

```python
MOE = z*SE
MOE
```

```python
lower = sample_mean - MOE
upper = sample_mean + MOE
lower, upper
```

So here we can see that our population mean is between that confidence interval of (21.54,30.09) Which is 27.62. Hence C.I Formula is correct.
