```python
df_financials.currency.value_counts()
```

```python
def standerdize_currency(curr):
    if curr == "$$" or curr == "Dollars":
        return "USD"
    else:
        return curr
df_financials = pd.read_excel("movies_db.xlsx","financials", converters={'currency': standerdize_currency})
df_financials.head()
```

```python
df_merged = pd.merge(df_movies,df_financials, on = "movie_id")
df_merged.head()
```

```python
df_merged.to_excel("movies_merged.xlsx", index = False)
```
