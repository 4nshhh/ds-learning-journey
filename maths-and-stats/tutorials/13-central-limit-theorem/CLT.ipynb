```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("bill_survey.csv")
df.head()
```

So Here it is city wise bill data so let's first plot a distribution of whole sample 

```python
sns.histplot(df.Bill,kde = True)
```

As we can see it is right skewed. Now let's take samples for each city and find mean of each city samples and then plot distribution of the sample means.

```python
per_city_mean = df.groupby("City").Bill.mean()
per_city_mean
```

```python
sns.histplot(per_city_mean.values, kde = True)
```

See the distribution of sample mean is normal.Even if the sample itself wasn't normally distributed.
