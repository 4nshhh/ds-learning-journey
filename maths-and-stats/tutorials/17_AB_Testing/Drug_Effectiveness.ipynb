```python
import numpy as np
import pandas as pd
import scipy.stats as st
```

We have sample data for two test group 
1. Control Group : Old Drug Data
2. Test Group : New Drug Data

* Null : Both drugs has same avg recovery time (control_mean == test_mean)
* Alternate: New drug has better recovery time (control_mean < test_mean) (One-tailed test)(Right tailed)

```python
control = pd.read_csv("control_group.csv")
print(control.shape)
control.head()
```

```python
test = pd.read_csv("test_group.csv")
print(test.shape)
test.head()
```

```python
control_mean = control.recovery_time_hrs.mean()
control_std = control.recovery_time_hrs.std()
control_size = control.shape[0]

control_mean, control_std, control_size
```

```python
test_mean = test.recovery_time_hrs.mean()
test_std = test.recovery_time_hrs.std()
test_size = test.shape[0]

test_mean, test_std, test_size
```

Let's find out Standerd error now using both the std

```python
SE = np.sqrt((np.square(control_std)/control_size) + (np.square(test_std)/test_size))
SE
```

```python
z_score = (control_mean - test_mean)/SE
z_score
```

```python
alpha = 0.05
z_critic = st.norm.ppf(1-alpha)
z_critic
```

```python
z_score, z_critic
```

Since z_score > z_critic, We can reject Null Hypothesis. Therefore, New drug is found to be more effective and has better recovery time then old drug.
