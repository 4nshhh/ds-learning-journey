```python
import time
def get_square(numbers):
    start = time.time()
    result = []
    for n in numbers:
        result.append(n*n)
    end = time.time()
    print(f"get_square took {(end - start)*1000} mil seconds")
    return result
```

```python
def get_cube(numbers):
    start = time.time()
    result = []
    for n in numbers:
        result.append(n*n*n)
    end = time.time()
    print(f"get_square took {(end - start)*1000} mili seconds")
    return result
```

```python
my_list = [1,2,3,4,5,6]
_ = get_square(range(100000))
```

```python
_ = get_cube(range(100000))
```

```python
# But here we are repeating the code to calculate the time taken 
# Here comes the concept of decorators in the picture!

def timer(func): # Here making the decorator 
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print(f"Function {func.__name__} took {(end - start)*1000} mili seconds")
        return result
    return wrapper
```

```python
# Now let's use it!

@timer
def get_square(numbers):
    result = []
    for n in numbers:
        result.append(n*n)
    return result
```

```python
@timer
def get_cube(numbers):
    result = []
    for n in numbers:
        result.append(n*n*n)
    return result
```

```python
_ = get_square(range(100000))
```

```python
_ = get_cube(range(100000))
```
