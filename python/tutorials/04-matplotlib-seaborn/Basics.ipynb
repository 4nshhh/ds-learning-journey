```python
from matplotlib import pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_excel("linechart.xlsx")
df
```

```python
plt.figure(figsize = (12,4))
plt.plot(df["Quarter"],df["Fridge"],label = "Fridge")
plt.plot(df["Quarter"],df["Dishwasher"],label = "Dishwasher")
plt.plot(df["Quarter"],df["Washing Machine"],label = "Washing Machine")
plt.xlabel("Quarter")
plt.ylabel("Revenue in million($)")
plt.legend()
plt.title("Sales Chart")
plt.show()
```

```python
total_sales = df[["Fridge","Dishwasher","Washing Machine"]].sum()
total_sales
```

```python
total_sales.index
```

```python
plt.pie(total_sales,labels= total_sales.index,explode = (0.1,0,0), shadow = True, autopct = "%1.1f%%",startangle= 140)

plt.show()
```

```python
df.plot(kind = "bar", x = "Quarter")
plt.ylabel("Revenue in mln($)")
plt.title("Sales Chart")
plt.xticks(rotation = 45)
plt.show()
```

```python
df_2 = df.set_index(["Quarter"])
df_2
```

```python
df_2.plot(kind = "bar")
plt.ylabel("Revenue in mln($)")
plt.title("Sales Chart")
plt.xticks(rotation = 45)
plt.show()
```
