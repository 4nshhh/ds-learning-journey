```python
import json
```

```python
# It's a JSON string
json_string = '''
{
    "name": "Maulik",
    "age": 30,
    "is_employee": true,
    "skills": ["Python", "Data Analysis", "Machine Learning"],
    "address": {
        "street": "123 Main St",
        "city": "Wonderland"
    }
}
'''
```

```python
type(json_string)
```

```python
# Let's Convert it into a python object (Dict)

obj = json.loads(json_string)
obj
```

```python
type(obj)
```

```python
# Now let's convert python obj to JSON String

data = {
    "name": "Bob",
    "age": 25,
    "is_employee": False,
    "skills": ["JavaScript", "Web Development"],
    "address": {
        "street": "456 Elm St",
        "city": "Springfield"
    }
}
```

```python
json_string =json.dumps(data)
json_string
```

```python
type(json_string)
```

```python
# How to write this JSON data to a file?

with open("output.json", "w") as f:
    json.dump(data,f, indent = 4)
```

```python
# How to read the data from JSON file?

with open("products.json", "r") as f:
    products = json.load(f)
products
```

```python
type(products)
```

```python
products[0]
```

```python
products[0]['price']
```

```python
# How to get only the title and author as python list object from this JSON string?
json_string = '''
{
    "books": [
        {"title": "The Magic Of Thinking Big", "author": "David J. Schwartz", "published": 1959},
        {"title": "Chanakya Neeti", "author": "B.K. Chaturvedi", "published": 2009},
        {"title": "How to Win Friends and Influence People", "author": "Dale Karnegie", "published": 1936}
    ],
    "status_code": 200
}
'''

data = json.loads(json_string)
data
```

```python
books =data["books"]
books
```

```python
book_info = [{'title' : book['title'], 'author' : book['author']} for book in books]
book_info
```

```python
# Error handelling in JSON File

# Malformed JSON string
json_string = '''
{
    "name": "Krishna",
    "age": 30
    "is_employee": true
}
'''

json.loads(json_string)
```

```python
try:
    json.loads(json_string)
except json.JSONDecodeError as e:
    print("Error Occurred:",e)
```
