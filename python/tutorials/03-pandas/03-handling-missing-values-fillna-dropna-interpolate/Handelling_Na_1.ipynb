```python
# Let's convert it into date data type 

df = pd.read_csv("weather_data.csv", parse_dates=["day"])
df
```

```python
# Now let's handle null vales

df.fillna(0)
```

```python
df.set_index("day",inplace = True)
df
```

```python
df.fillna({
    "temperature" : df.temperature.mean(),
    "windspeed" : df.windspeed.mean(),
    "event" : "No Event"
})
```

```python
df.ffill()
```

```python
df.bfill()
```

```python
df.interpolate()
```

```python
df.dropna()
```

```python
df.dropna(how ="all")
```
