```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heights_with_outliers.csv")
df.head()
```

```python
mean = df.height.mean()
std = df.height.std()
mean,std
```

```python
sns.histplot(df.height, kde = True)
plt.axvline(mean, color = 'black',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - std, color = 'blue',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + std, color = 'blue',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - 2*std, color = 'green',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + 2*std, color = 'green',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - 3*std, color = 'yellow',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + 3*std, color = 'yellow',linestyle = 'dashed', linewidth = 1)
```

```python
lower = round(mean - 3*std,2)
higher = round(mean + 3*std,2)
lower,higher
```

```python
df[(df.height < lower) | (df.height > higher)]
```

This values are outliers so let's remove them

```python
filtered_df = df[(df.height > lower) & (df.height < higher)]
filtered_df.shape
```

```python
sns.histplot(filtered_df.height, kde = True)
plt.axvline(mean, color = 'black',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - std, color = 'blue',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + std, color = 'blue',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - 2*std, color = 'green',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + 2*std, color = 'green',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - 3*std, color = 'yellow',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + 3*std, color = 'yellow',linestyle = 'dashed', linewidth = 1)
```

So this is the data now without having any outliers
