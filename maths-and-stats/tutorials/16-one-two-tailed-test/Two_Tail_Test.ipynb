# Two Tail

```python
import pandas as pd
import numpy as np
import scipy.stats as st
```

**Null Hypothesis:** House Inflation is 10%

**Alternate Hypothesis:** House Inflation is not 10%

```python
sample = pd.read_csv("house_price_increase.csv")
sample.head()
```

```python
sample.shape
```

For z test we need to find z value using formula. Which is z = (sample mean - popultion mean)/S.E

Suppose we have the population mean and std available below

```python
population_mean = 10
population_std = 4
```

```python
sample_mean = sample.price_increase_pct.mean()
sample_mean
```

```python
# Standerd Error

SE = population_std/np.sqrt(100) # Sample size is 100
SE
```

```python
z_score = (sample_mean-population_mean)/SE
abs(z_score) # It can be -ve too
```

```python
# Now at 5% significance level 
alpha = 0.05

z_critic = st.norm.ppf(1-alpha/2) # Becuase it's a Two Tail test so alpha/2
z_critic
```

Since z_score > z_critic, We Will reject the NULL Hypothesis. This means the alternate hypothesis becomes true which means the inflation is not 10%

### Z test Using p-value

So here the z score is 2.5

```python
p_value = 2*(1 - st.norm.cdf(abs(z_score))) # We have to multiply by 2 to get p_value
p_value
```

```python
alpha = 0.05
(p_value, alpha)
```

As we can see that p_value is less than alpha so we will reject Null Hyptothesis
