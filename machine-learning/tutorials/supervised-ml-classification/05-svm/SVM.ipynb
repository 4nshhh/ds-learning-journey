```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
df = pd.read_excel("Raisin_Dataset.xlsx")
df.head()
```

```python
# Define Features and Target vars
X = df.drop("Class", axis = 1)
y = df["Class"]
```

```python
# Create a train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

## Using rbf kernal

```python
# Train model on SVC (Support Vector Classfier)
model = SVC(kernel="rbf")
model.fit(X_train, y_train)

# Test the model on test data set
y_pred = model.predict(X_test)
```

```python
cr = classification_report(y_test, y_pred)
print(cr)
```

## Using Linear kernal

```python
# Now let's check how the model performs if we use linear kernal instead of rbf(Default)
linear_model = SVC(kernel="linear")
linear_model.fit(X_train, y_train)

y_pred_linear = linear_model.predict(X_test)

cr_linear = classification_report(y_test, y_pred_linear)
print(cr_linear)
```

Oh it's giving better preformance then our previous model should we use this? NO

We first have to check that how many iterations both the model having to train itself.

```python
model.n_iter_, linear_model.n_iter_
```

As we can see that the linear model is taking so many more iterations to give this performance and Higher iteration = Higher computational power = Higher cost to train the model

We can improve the rbf model performance by scaling the features into single scale. So model can learn patterns well.
