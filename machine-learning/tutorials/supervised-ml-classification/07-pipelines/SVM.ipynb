```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
df = pd.read_excel("Raisin_Dataset.xlsx")
df.head()
```

```python
# Define Features and Target vars
X = df.drop("Class", axis = 1)
y = df["Class"]
```

```python
# Create a train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

## Using rbf kernal

```python
# Train model on SVC (Support Vector Classfier)
model = SVC(kernel="rbf")
model.fit(X_train, y_train)

# Test the model on test data set
y_pred = model.predict(X_test)
```

```python
cr = classification_report(y_test, y_pred)
print(cr)
```

We can improve the rbf model performance by scaling the features into single scale. So model can learn patterns well.

## Using Standerd Scaler:

```python
from sklearn.preprocessing import StandardScaler, minmax_scale # We can also use min-max scaler

scaler = StandardScaler()

# Fit the scaler on features
scaler.fit(X)

# Transform the train and test data of features to scaled data
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train_scaled
```

## Model Traning on scaled data

```python
# Train model on SVC (Support Vector Classfier)
model = SVC(kernel="rbf")
model.fit(X_train_scaled, y_train)

# Test the model on test data set
y_pred = model.predict(X_test_scaled)
```

```python
cr = classification_report(y_test, y_pred)
print(cr)

model.n_iter_
```

As we can see there is preety good improvment in our model after scaling.

But How can we reduce lengthy process of scaling and then training to just few lines of code? Using sklearn pipelines!

## Sklearn Pipelines

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('svc', SVC())
])
```

```python
# Fit the data into pipeline just like we do with model
pipeline.fit(X_train, y_train)

y_pred = pipeline.predict(X_test)

cr = classification_report(y_test, y_pred)
print(cr)

model.n_iter_
```

So here we done above whole process just in few lines using pipeline method.
