```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.datasets import make_classification
from sklearn.model_selection import KFold, train_test_split, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

We create K-folds to know avg performance of our model. We use each fold once as test dataset and all other folds are used as train set

```python
# Creating synthetic Data set
X, y = make_classification(
    n_features=10,
    n_samples=1000,
    n_informative=8,
    n_redundant=2,
    n_repeated=0,
    n_classes=2,
    random_state=42
)
```

```python
# Creating 5 folds
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

Let's get the Cross validation scores now for different algorithms

### Logistic Regression

```python
score_lr = cross_val_score(LogisticRegression(), X, y, cv = kf)
score_lr
```

```python
# We will judge performance by the avg score
np.mean(score_lr)
```

### Decision Tree Classifier

```python
score_dt = cross_val_score(DecisionTreeClassifier(), X, y, cv = kf)
score_dt
```

```python
# We will judge performance by the avg score
np.mean(score_dt)
```

### Random Forst

```python
score_rf = cross_val_score(RandomForestClassifier(), X, y, cv = kf)
score_rf
```

```python
# We will judge performance by the avg score
np.mean(score_rf)
```

Basically cross_val_score and cross_validate do the same job but the only difference is,
* cross_val_score = quick result
* cross_validate = full report

If u want train vs test comparison (overfitting check)

You want multiple metrics like:
* accuracy
* precision
* recall

Then use cross_validate over cross_val_score

```python
from sklearn.model_selection import cross_validate

cross_validate(RandomForestClassifier(), X, y, cv = kf, scoring=['roc_auc'] )
```

If you only use cross_val_score, you might miss:

* overfitting (no train score)
* performance details

* That’s why professionals often prefer cross_validate
