```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heights.csv")
df.head()
```

```python
mean = round(df.height.mean(),2)
std = round(df.height.std(),2)
mean, std
```

```python
sns.histplot(df.height, kde = True)
plt.axvline(mean, color = 'black',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - std, color = 'blue',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + std, color = 'blue',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - 2*std, color = 'green',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + 2*std, color = 'green',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean - 3*std, color = 'yellow',linestyle = 'dashed', linewidth = 1)
plt.axvline(mean + 3*std, color = 'yellow',linestyle = 'dashed', linewidth = 1)
```
