```python
# URL of the end point 

url = "https://jsonplaceholder.typicode.com/posts"

# Data to send in the post request

data = {
    'userId': 2,
    'id': 101,
    'title': 'bro',
    'body': 'yo'  
}
```

```python
response = requests.post(url, json = data)
response.status_code
```

201 means the post request is successful!

```python
# Getting the posted data

if response.status_code == 201:
    created_post = response.json()
    print(created_post)
else:
    print("Failed to create post")
```

#### Common Status Codes

1. 200: OK (The request was successful).
1. 201: Created (The resource was successfully created).
1. 400: Bad Request (The server could not understand the request).
1. 401: Unauthorized (Authentication is required).
1. 404: Not Found (The requested resource could not be found).
1. 500: Internal Server Error (The server encountered an error).

```python
# URL of the API endpoint
url = "https://jsonplaceholder.typicode.com/posts/1"

# Sending a GET request
response = requests.get(url)

# Handling different status codes
if response.status_code == 200:
    print("Request was successful!")
    print(response.json())
elif response.status_code == 404:
    print("Resource not found.")
else:
    print(f"Failed with status code: {response.status_code}")
```

```python
# Send query param within our GET request

url = "https://jsonplaceholder.typicode.com/posts"

# Giving the query param id as 1 so the data with id : 1 will be only fetched in our get request
param = {
    'id' : 1
}

response  = requests.get(url, params= param)

if response.status_code == 200:
    print("Request was successful!")
    print(response.json())
elif response.status_code == 404:
    print("Resource not found.")
else:
    print(f"Failed with status code: {response.status_code}")
```

```python
import os
from dotenv import load_dotenv
import requests

# Load .env file
load_dotenv()

# Get API key from .env
api_key = os.getenv("API_KEY")

url = "https://api.api-ninjas.com/v1/bitcoin"

headers = {
    "X-Api-Key": api_key
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    print("Request was successful!")
    print(response.json())
else:
    print("Error:", response.status_code)
```
